## 5. Data Cleaning

The TotalCharges column contains blank values which are stored
as text in the original dataset.

These values need to be converted into numeric format.
The blank entries are checked against customer tenure before
deciding how to handle them.

```python
# Converting TotalCharges into numeric format

df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
)

print("Missing values after conversion:")
print(df['TotalCharges'].isnull().sum())

# Checking records where TotalCharges is missing

df[df['TotalCharges'].isnull()][
    ['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']
]

# Checking the tenure of customers with missing TotalCharges

df[df['TotalCharges'].isnull()]['tenure'].value_counts()

# Replacing missing TotalCharges with 0 for zero-tenure customers

zero_tenure_missing = (
    df['TotalCharges'].isnull() &
    (df['tenure'] == 0)
)

df.loc[zero_tenure_missing, 'TotalCharges'] = 0

print("Remaining missing TotalCharges:",
      df['TotalCharges'].isnull().sum())

# Removing duplicate records, if any

df = df.drop_duplicates()

print("Shape after duplicate check:", df.shape)
```

## 6. Feature Engineering

New analytical columns are created to make customer segmentation
and churn analysis easier.

These features will also be used in the SQL analysis and
machine learning stages.

```python
# Converting Churn into numerical values

df['ChurnFlag'] = df['Churn'].map({
    'Yes': 1,
    'No': 0
})

df[['Churn', 'ChurnFlag']].head()

# Grouping customers based on their tenure

df['TenureBucket'] = pd.cut(
    df['tenure'],
    bins=[-1, 12, 24, 48, 72],
    labels=['0-12', '13-24', '25-48', '49-72']
)

df[['tenure', 'TenureBucket']].head()

# Counting additional services used by each customer

addon_columns = [
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport'
]

df['AddonCount'] = (
    df[addon_columns] == 'Yes'
).sum(axis=1)

df[['customerID', 'AddonCount']].head()

# Counting total subscribed services

df['ServiceCount'] = (
    (df['PhoneService'] == 'Yes').astype(int) +
    (df['InternetService'] != 'No').astype(int) +
    (df['OnlineSecurity'] == 'Yes').astype(int) +
    (df['OnlineBackup'] == 'Yes').astype(int) +
    (df['DeviceProtection'] == 'Yes').astype(int) +
    (df['TechSupport'] == 'Yes').astype(int) +
    (df['StreamingTV'] == 'Yes').astype(int) +
    (df['StreamingMovies'] == 'Yes').astype(int)
)

df[['customerID', 'ServiceCount']].head()

# Categorising customers based on MonthlyCharges

df['ChargeBand'] = pd.cut(
    df['MonthlyCharges'],
    bins=[-np.inf, 35, 70, 90, np.inf],
    labels=['<35', '35-70', '70-90', '90+'],
    right=False
)

df[['MonthlyCharges', 'ChargeBand']].head()
```
